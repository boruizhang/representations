```python
%pip install tiktoken transformers
import tiktoken
from transformers import BertTokenizer

# GPT-4 tokenizer (BPE)
gpt_enc = tiktoken.encoding_for_model("gpt-4")

# BERT tokenizer (WordPiece)
bert_tok = BertTokenizer.from_pretrained("bert-base-uncased")

```

```python
words = [
    "undoable", # ambiguous word
    "pneumonoultramicroscopicsilicovolcanoconiosis", # long word
    "飞机是今天晚上七点钟准时降落在北京首都国际机场的",# Chinese - the flight will land on time at the Beijing Capital Airport at 7.
    "걱정",           # Korean - worry
    "食べさせられた",   # Japanese causative-passive - was made to eat (eat + cause + passive + past)
    "evlerimizden",   # Turkish agglutinative - from our houses (house + plural + possession + ablative case)
    "Naturwissenschaften",  # German compound - natural sciences
    "supercalifragilistic", # not a "word"? playful. - super good.
]

#Loop through words
for word in words:
    gpt_tokens = [
        gpt_enc.decode_single_token_bytes(t).decode("utf-8", errors="replace")
        for t in gpt_enc.encode(word)
    ]

    bert_tokens = bert_tok.tokenize(word)

    print(f"\n'{word}'")
    print(f"  GPT (BPE):      {gpt_tokens}  ({len(gpt_tokens)} tokens)")
    print(f"  BERT (WordPiece): {bert_tokens}  ({len(bert_tokens)} tokens)")
```
