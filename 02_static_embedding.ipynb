#GloVe (Global Vectors for Word Representation) which functions similarly to Word2Vec, as a static embedding. Explore the provided examples.

```python
#package setting up
%pip install gensim #(Generate Similarities) https://pypi.org/project/gensim/
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('ggplot')
```

```python
# getting pretrained model
from sklearn.decomposition import PCA
import gensim.downloader as api # getting ready-to-go models
from gensim.models import KeyedVectors # stores dictionary (words) --> vector mapping
model = api.load('glove-wiki-gigaword-100') # built in 2014, around 6 billion tokens (Wikipedia and Gigaword news text)
print(type(model))
```

```python
model['jedi']
```

```python
model['yoda']
```

```python
model.distance('jedi','yoda')
```

```python
model.similarity('jedi', 'yoda')
```

```python
model.most_similar('jedi')
```

```python
model.most_similar(negative = 'study')
```

```python
model.doesnt_match(['milk','jedi','yogurt','car'])
```

```python
# Semantic embedding analogy
result = model.most_similar(positive = ['woman','king'], negative = ['man'])
print("{}: {:.4f}".format(*result[0]))
```

```python
def analogy(x1,x2,y1):
  result = model.most_similar(positive = [y1,x2],negative = [x1])
  return result[0][0]
```

```python
analogy('man','king','woman')
```

```python
analogy('king','man','queen')
```
